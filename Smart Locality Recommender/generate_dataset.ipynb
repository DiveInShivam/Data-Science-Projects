```python
import pandas as pd
import numpy as np
n = 1000
#  Localities list (added 5 more)
localities = [
    "Rohini", "Saket", "Dwarka", "Laxmi Nagar", "Karol Bagh",
    "Janakpuri", "Pitampura", "Rajouri Garden", "Vasant Kunj",
    "Mukherjee Nagar", "Kamla Nagar", "GTB Nagar",
    "Mayur Vihar", "Ashok Vihar", "Tilak Nagar", "Malviya Nagar", "Okhla"
]
# (Optional but better) Base rent for realism
locality_base_rent = {
    "Rohini": 8000, "Saket": 15000, "Dwarka": 10000, "Laxmi Nagar": 9000,
    "Karol Bagh": 14000, "Janakpuri": 11000, "Pitampura": 12000,
    "Rajouri Garden": 13000, "Vasant Kunj": 16000,
    "Mukherjee Nagar": 8500, "Kamla Nagar": 10500, "GTB Nagar": 9500,
    "Mayur Vihar": 10000, "Ashok Vihar": 11000,
    "Tilak Nagar": 9500, "Malviya Nagar": 14000, "Okhla": 8000
}
# Generate locality column
locality_col = np.random.choice(localities, n)
# Generate realistic rent based on locality
avg_rent = [
    locality_base_rent[loc] + np.random.randint(-2000, 2000)
    for loc in locality_col
]
data = {
    "locality": locality_col,
    "city": ["Delhi"] * n,
    "avg_rent": avg_rent,
    "distance_to_college": np.round(np.random.uniform(0.5, 20, n), 2),
    "metro_distance": np.round(np.random.uniform(0.1, 5, n), 2),
    "safety_score": np.random.randint(1, 10, n),
    "food_score": np.random.randint(1, 10, n),
    "hospital_count": np.random.randint(1, 20, n),
    "mall_count": np.random.randint(0, 15, n),
    "public_transport_score": np.random.randint(1, 10, n),
    "pollution_index": np.random.randint(50, 300, n),
    "internet_speed": np.random.randint(10, 200, n),
    "student_rating": np.round(np.random.uniform(1, 5, n), 1)
}
df = pd.DataFrame(data)
df.to_csv("delhi_smart_locality.csv", index=False)
print(" Dataset created successfully!")
```
